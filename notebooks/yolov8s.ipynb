```python
import os
from ultralytics import YOLO

def train_yolov8s_model(
    data_yaml_path,
    model_yaml='yolov8s.yaml',        # YOLOv8 small model
    epochs=350,
    imgsz=640,
    project_name='/kaggle/working/yolo8_training_runs'
):
    """
    Trains a YOLOv8s model from scratch using the ultralytics library.
    """

    print(" Starting YOLOv8s model training from scratch...")

    # Check if dataset YAML exists
    if not os.path.exists(data_yaml_path):
        print(f" Error: The data.yaml file '{data_yaml_path}' was not found.")
        return

    try:
        # Initialize YOLOv8s model (architecture only)
        model = YOLO(model_yaml)

        # Ensure all parameters are trainable
        for name, param in model.model.named_parameters():
            param.requires_grad = True
            if not param.requires_grad:
                print(f" {name} is frozen")

        # Start training
        results = model.train(
            data=data_yaml_path,
            epochs=epochs,
            imgsz=imgsz,
            project=project_name,
            save_period=10,      # Save weights every 10 epochs
            pretrained=False,    # True = load pretrained COCO weights; False = from scratch
            patience=20,
            freeze=0             # No layers frozen
        )

        print("\n Training completed successfully!")
        print(f" Results saved in: {results.save_dir}")

    except Exception as e:
        print(f"Error during training: {e}")
        print("Please ensure Ultralytics is installed and GPU (CUDA) is available.")

# Example usage
if __name__ == "__main__":
    data_yaml_file = '/kaggle/working/data_yolov11.yaml'
    train_yolov8s_model(data_yaml_file)
```
