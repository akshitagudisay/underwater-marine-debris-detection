```python
import os
from ultralytics import YOLO

def train_yolo9s_model(
    data_yaml_path,
    model_yaml='yolov9s.yaml',            # YOLOv9s architecture
    epochs=350,                           # Same as original
    imgsz=640,
    project_name='/kaggle/working/yolo9s_training_runs'
):
    """
    Trains a YOLOv9s model from scratch using the Ultralytics library.
    """

    print("Starting YOLOv9s model training from scratch...")

    if not os.path.exists(data_yaml_path):
        print(f"Error: The data.yaml file '{data_yaml_path}' was not found.")
        return

    try:
        # Load the YOLOv9s model architecture from YAML
        model = YOLO(model_yaml)

        # Make sure all weights are trainable
        for name, param in model.model.named_parameters():
            param.requires_grad = True
            if not param.requires_grad:
                print(name, "is frozen")

        # Train the model with early stopping & save every 10 epochs
        results = model.train(
            data=data_yaml_path,
            epochs=epochs,
            imgsz=imgsz,
            project=project_name,
            save_period=10,
            pretrained=False,  # train from scratch
            patience=20,
            freeze=0
        )

        print("\nTraining completed successfully!")
        print("Training results are saved in the 'runs/detect' directory.")

    except Exception as e:
        print(f"An error occurred during training: {e}")
        print("Please ensure you have ultralytics and a CUDA-enabled GPU set up correctly.")


# Example usage
if __name__ == "__main__":
    data_yaml_file = '/kaggle/working/data_yolov11.yaml'
    train_yolo9s_model(data_yaml_file)
```

```python
# =========================
# TEST SET EVALUATION
# =========================

metrics = model.val(
    data=data_yaml_dst,    
    split="test"
)

# Extract test metrics
precision = metrics.box.mp
recall = metrics.box.mr
map50 = metrics.box.map50
map5095 = metrics.box.map

# F1-score from aggregate Precision and Recall
f1 = (
    2 * precision * recall / (precision + recall)
    if (precision + recall) > 0
    else 0
)

# Display results
print("\n" + "=" * 45)
print("TEST SET RESULTS")
print("=" * 45)
print(f"Precision    : {precision:.4f}")
print(f"Recall       : {recall:.4f}")
print(f"F1-Score     : {f1:.4f}")
print(f"mAP@50       : {map50:.4f}")
print(f"mAP@50-95    : {map5095:.4f}")
print("=" * 45)
```
